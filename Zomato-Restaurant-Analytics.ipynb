# **Project Summary -**

Zomato Restaurant Analytics is an end-to-end machine learning and analytics project designed to analyze restaurant business patterns and segment restaurants into meaningful clusters based on customer ratings, pricing, reviews, sentiment, cuisine, and collection-related attributes. The project combines data analytics, machine learning, and web application development to transform restaurant-level data into actionable business insights.

The primary objective of the project is to identify distinct restaurant segments using unsupervised machine learning, specifically K-Means clustering. Restaurant characteristics such as average rating, average cost for two people, review count, average pictures per review, positive sentiment, negative sentiment, neutral sentiment, cuisine encoding, and collection encoding are used as model features. These features allow the system to discover similarities and differences between restaurants without requiring predefined business categories.

The analytical component provides insights into restaurant performance and customer perception. The dataset contains 12,450 restaurants, with an overall average rating of approximately 4.02, average cost of 840, and positive sentiment of 64.7%. The clustering analysis identifies six major business segments: Standard Restaurants, Premium Highly Rated, Budget Friendly, Luxury Dining, Quick Bites, and Cafe & Bakery. These segments provide a structured view of different restaurant business profiles.

The project also evaluates the characteristics of each cluster through average ratings, average costs, and sentiment patterns. For example, Premium and Luxury restaurants demonstrate higher average ratings and pricing, while Budget and Quick-service segments represent more cost-conscious restaurant categories. Sentiment analysis further helps understand customer satisfaction patterns across different restaurant types.

To make the machine learning solution accessible, the trained model is integrated into a web-based application. The frontend is developed using React, TypeScript, Vite, and supporting visualization libraries, while the backend exposes machine learning functionality through a FastAPI-based API. Users can enter restaurant metrics through the prediction interface and obtain the predicted restaurant cluster.

The application is deployed using Vercel, with the frontend and backend configured for cloud-based access. Environment variables are used to connect the frontend with the deployed API. The final solution therefore demonstrates the complete machine learning lifecycle, from data processing and exploratory analysis to clustering, model serving, frontend integration, deployment, and interactive prediction.

Overall, Zomato Restaurant Analytics provides a practical business intelligence solution that demonstrates how machine learning can be applied to restaurant data for market segmentation, customer sentiment analysis, performance comparison, and data-driven decision-making.

#### Chart - 14 - Correlation Heatmap

````python
# Chart - 14 - Correlation Heatmap

df.columns = df.columns.astype(str).str.strip()

candidate_cols = [
    "Average Rating",
    "Average Cost for Two",
    "Review Count",
    "Average Pictures per Review",
    "Positive Sentiment",
    "Negative Sentiment",
    "Neutral Sentiment",
    "Cuisine (Encoded)",
    "Collection (Encoded)"
]

available_cols = [
    col for col in candidate_cols
    if col in df.columns
]

corr_df = df[available_cols].copy()

for col in corr_df.columns:
    corr_df[col] = pd.to_numeric(
        corr_df[col],
        errors="coerce"
    )

corr_df = corr_df.dropna(
    axis=1,
    how="all"
)

if corr_df.shape[1] < 2:
    print("Not enough numeric columns available for a correlation heatmap.")
    print("Available columns in dataset:")
    print(df.columns.tolist())

else:
    corr_matrix = corr_df.corr()

    corr_matrix = corr_matrix.dropna(
        axis=0,
        how="all"
    ).dropna(
        axis=1,
        how="all"
    )

    plt.figure(figsize=(12, 8))

    sns.heatmap(
        corr_matrix,
        annot=True,
        cmap="coolwarm",
        fmt=".2f",
        linewidths=0.5
    )

    plt.title("Correlation Between Restaurant Variables")
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)

    plt.tight_layout()
    plt.show()
````

##### 1. Why did you pick the specific chart?

I selected a correlation heatmap because it clearly shows the strength and direction of relationships between numerical restaurant variables. It helps identify which variables are positively or negatively related and makes it easier to compare multiple relationships at the same time.

##### 2. What is/are the insight(s) found from the chart?

The correlation heatmap shows the relationships among the numerical variables in the restaurant dataset. Positive correlation indicates that two variables tend to increase together, while negative correlation indicates that one tends to decrease as the other increases. ## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Based on the chart experiments, three hypotheses were developed. The first tests whether average cost for two is significantly related to average restaurant rating. The second tests whether positive customer sentiment is significantly associated with restaurant ratings. The third tests whether positive and negative customer sentiment differ significantly. Statistical hypothesis tests will be performed to determine whether the null hypotheses can be rejected and to draw final conclusions from the dataset.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀): There is no significant relationship between the average cost for two people and the average restaurant rating.

Alternative Hypothesis (H₁): There is a significant relationship between the average cost for two people and the average restaurant rating.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test - Hypothesis 1
# Pearson Correlation Test

from scipy.stats import pearsonr
import pandas as pd

df.columns = df.columns.astype(str).str.strip()

rating_candidates = [
    "Average Rating",
    "Rating",
    "Ratings",
    "Aggregate rating"
]

cost_candidates = [
    "Average Cost for Two",
    "Cost for Two",
    "Cost"
]

rating_col = next(
    (col for col in rating_candidates if col in df.columns),
    None
)

cost_col = next(
    (col for col in cost_candidates if col in df.columns),
    None
)

if rating_col is None or cost_col is None:
    print("Required columns were not found.")
    print("\nAvailable columns:")
    print(df.columns.tolist())

else:
    test_df = df[[cost_col, rating_col]].copy()

    test_df[cost_col] = pd.to_numeric(
        test_df[cost_col],
        errors="coerce"
    )

    test_df[rating_col] = pd.to_numeric(
        test_df[rating_col],
        errors="coerce"
    )

    test_df = test_df.dropna()

    correlation, p_value = pearsonr(
        test_df[cost_col],
        test_df[rating_col]
    )

    print("Statistical Test: Pearson Correlation Test")
    print("Variable 1:", cost_col)
    print("Variable 2:", rating_col)
    print("Sample Size:", len(test_df))
    print("Correlation coefficient:", round(correlation, 4))
    print("P-value:", round(p_value, 6))

    alpha = 0.05

    if p_value < alpha:
        print("\nDecision: Reject the Null Hypothesis (H0).")
        print("There is a statistically significant relationship between the two variables.")
    else:
        print("\nDecision: Fail to Reject the Null Hypothesis (H0).")
        print("There is not enough statistical evidence of a significant relationship.")
````

##### Which statistical test have you done to obtain P-Value?

Pearson Correlation Test was used to obtain the p-value.

##### Why did you choose the specific statistical test?

Pearson correlation was chosen because the hypothesis examines whether there is a significant linear relationship between two numerical variables: Average Cost for Two and Average Rating. The test provides a correlation coefficient and a p-value to determine whether the observed relationship is statistically significant.

H₁ (Alternative Hypothesis): There is a significant relationship between Positive Sentiment and Average Rating.

#### 2. Perform an appropriate statistical test.

````python
# Hypothetical Statement - 2
# Pearson Correlation Test:
# Positive Sentiment vs Average Rating

from scipy.stats import pearsonr
import pandas as pd

df.columns = df.columns.astype(str).str.strip()

sentiment_candidates = [
    "Positive Sentiment",
    "Positive_Sentiment",
    "Positive"
]

rating_candidates = [
    "Average Rating",
    "Average_Rating",
    "Rating",
    "Ratings",
    "Aggregate rating"
]

sentiment_col = next(
    (c for c in sentiment_candidates if c in df.columns),
    None
)

rating_col = next(
    (c for c in rating_candidates if c in df.columns),
    None
)

if sentiment_col is None or rating_col is None:

    print("Required columns not found.")
    print("\nAvailable columns:")
    print(df.columns.tolist())

else:

    test_df = df[[sentiment_col, rating_col]].copy()

    test_df[sentiment_col] = pd.to_numeric(
        test_df[sentiment_col],
        errors="coerce"
    )

    test_df[rating_col] = pd.to_numeric(
        test_df[rating_col],
        errors="coerce"
    )

    test_df = test_df.dropna()

    correlation, p_value = pearsonr(
        test_df[sentiment_col],
        test_df[rating_col]
    )

    print("Statistical Test: Pearson Correlation Test")
    print("Variable 1:", sentiment_col)
    print("Variable 2:", rating_col)
    print("Sample Size:", len(test_df))
    print("Correlation coefficient:", round(correlation, 4))
    print("P-value:", round(p_value, 6))

    alpha = 0.05

    if p_value < alpha:
        print("\nDecision: Reject H0.")
        print("There is a statistically significant relationship.")
    else:
        print("\nDecision: Fail to Reject H0.")
        print("There is not enough evidence of a statistically significant relationship.")
````

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

H₀ (Null Hypothesis): There is no significant difference between Positive Sentiment and Negative Sentiment.

H₁ (Alternative Hypothesis): There is a significant difference between Positive Sentiment and Negative Sentiment.

#### 2. Perform an appropriate statistical test.

````python
# Hypothetical Statement - 3
# Paired Samples t-test:
# Positive Sentiment vs Negative Sentiment

from scipy.stats import ttest_rel
import pandas as pd

df.columns = df.columns.astype(str).str.strip()

positive_candidates = [
    "Positive Sentiment",
    "Positive_Sentiment",
    "Positive"
]

negative_candidates = [
    "Negative Sentiment",
    "Negative_Sentiment",
    "Negative"
]

positive_col = next(
    (c for c in positive_candidates if c in df.columns),
    None
)

negative_col = next(
    (c for c in negative_candidates if c in df.columns),
    None
)

if positive_col is None or negative_col is None:

    print("Required sentiment columns were not found.")
    print("\nAvailable columns:")
    print(df.columns.tolist())

else:

    test_df = df[[positive_col, negative_col]].copy()

    test_df[positive_col] = pd.to_numeric(
        test_df[positive_col],
        errors="coerce"
    )

    test_df[negative_col] = pd.to_numeric(
        test_df[negative_col],
        errors="coerce"
    )

    test_df = test_df.dropna()

    t_stat, p_value = ttest_rel(
        test_df[positive_col],
        test_df[negative_col]
    )

    print("Statistical Test: Paired Samples t-test")
    print("Positive Sentiment:", positive_col)
    print("Negative Sentiment:", negative_col)
    print("Sample Size:", len(test_df))
    print("Mean Positive Sentiment:",
          round(test_df[positive_col].mean(), 4))
    print("Mean Negative Sentiment:",
          round(test_df[negative_col].mean(), 4))
    print("T-statistic:", round(t_stat, 4))
    print("P-value:", round(p_value, 6))

    alpha = 0.05

    if p_value < alpha:
        print("\nDecision: Reject H0.")
        print("There is a statistically significant difference "
              "between Positive and Negative Sentiment.")
    else:
        print("\nDecision: Fail to Reject H0.")
        print("There is not enough evidence of a statistically "
              "significant difference.")
````

##### Which statistical test have you done to obtain P-Value?

Paired Samples t-test

##### Why did you choose the specific statistical test?

I chose the Paired Samples t-test because Positive Sentiment and Negative Sentiment are measured for the same observations. The test compares the two related measurements and determines whether their mean difference is statistically significant.

````python
# ML Model - 1 Implementation
# Random Forest Regression to predict Restaurant Cost

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


ml_df = df.copy()

ml_df["Cost"] = pd.to_numeric(
    ml_df["Cost"],
    errors="coerce"
)

ml_df = ml_df.dropna(subset=["Cost"])


numeric_cols = ml_df.select_dtypes(include=np.number).columns.tolist()

feature_cols = [
    col for col in numeric_cols
    if col != "Cost"
]

feature_cols = [
    col for col in feature_cols
    if ml_df[col].nunique(dropna=True) > 1
]

X = ml_df[feature_cols]
y = ml_df["Cost"]

print("Target variable: Cost")
print("\nNumber of features:", len(feature_cols))
print("\nFeatures used:")
print(feature_cols)

print("\nDataset shape:", X.shape)


X = X.fillna(0)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining rows:", len(X_train))
print("Testing rows:", len(X_test))


model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)


y_pred = model.predict(X_test)


mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("\n========== MODEL PERFORMANCE ==========")
print("MAE :", round(mae, 4))
print("RMSE:", round(rmse, 4))
print("R²  :", round(r2, 4))

metrics = ["MAE", "RMSE", "R²"]
values = [mae, rmse, r2]

plt.figure(figsize=(8, 5))
plt.bar(metrics, values)

plt.title("Random Forest Regression - Evaluation Metrics")
plt.xlabel("Evaluation Metric")
plt.ylabel("Score")

plt.tight_layout()
plt.show()
````

````python
# Cross-Validation & Hyperparameter Tuning
# Random Forest + GridSearchCV

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

rf_model = RandomForestRegressor(
    random_state=42
)

param_grid = {
    "n_estimators": [50, 100, 150],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}

grid_search = GridSearchCV(
    estimator=rf_model,
    param_grid=param_grid,
    cv=5,
    scoring="neg_mean_squared_error",
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best Hyperparameters:")
print(grid_search.best_params_)

best_model = grid_search.best_estimator_

y_pred_tuned = best_model.predict(X_test)

tuned_mae = mean_absolute_error(y_test, y_pred_tuned)

tuned_rmse = np.sqrt(
    mean_squared_error(y_test, y_pred_tuned)
)

tuned_r2 = r2_score(
    y_test,
    y_pred_tuned
)

print("\nTuned Model Evaluation")
print("----------------------")
print("MAE  :", round(tuned_mae, 4))
print("RMSE :", round(tuned_rmse, 4))
print("R²   :", round(tuned_r2, 4))
````

#### 3. Explain each evaluation metric's indication towards business and the business impact pf the ML model used.

MAE (Mean Absolute Error): Shows the average absolute prediction error. A lower MAE means predictions are closer to the actual values, which can help the business make more accurate decisions.

RMSE (Root Mean Squared Error): Measures prediction error while giving more importance to larger errors. A lower RMSE indicates fewer large prediction mistakes, reducing the risk of costly incorrect decisions.

R² (R-squared): Shows how much variation in the target variable is explained by the model. A higher R² indicates better explanatory power and can provide more reliable predictions for business planning and decision-making.
